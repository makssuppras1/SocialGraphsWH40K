### Degree distribution
The degree distribution shows a clear right-skewed pattern, which means most of the rock artists have only a few connections to other artists. However, there are a small number of artists who have a lot of connections, standing out as the most highly connected ones in the network. This type of distribution is typical of scale-free network where most of the nodes are not very well connected, but a small group of nodes, the "hubs", have a bigger number of connections. These highly connected artists act as central points in the network, connecting many others.

Looking at the log-log plot, it's clear that as the degree increases, the number of artists with that many connections drops off quickly. This shows that only a few artists are the big hubs, and most others are less connected, supporting the idea that the network follows the scale-free structure. The common structure in real-world networks like the internet, social networks, and in this case.

**Degree vs. Betweenness Centrality (left plot):** The higher an artist’s degree centrality (more connections), the more likely they are to have higher betweenness centrality (act as a bridge in the network). This shows that the more connected artists are also important in connecting different parts of the network.

**Degree vs. Eigenvector Centrality (right plot):** There's an even stronger link between degree centrality and eigenvector centrality. Artists with more connections tend to also be connected to other important, well-connected artists. This means that highly connected artists are usually connected to other influential artists as well.

Overall, these plots suggest that artists with more connections are important in the network in multiple ways, not just because they’re well-connected, but also because they connect to other key players in the network.

### Degree Assortativity 

WH40K: $(r = -0.025)$

**Assortativity coefficient** measures whether nodes with similar degrees tend to connect to each other:
- **Positive assortativity (> 0):** High-degree nodes connect to other high-degree nodes (assortative network)
- **Negative assortativity (< 0):** High-degree nodes connect to low-degree nodes (disassortative network)
- **Zero:** No preference (neutral)

**Interpretation:**
- Assortative networks: Hubs connect to other hubs, creating a "rich club" structure
- Disassortative networks: Hubs connect to many peripheral nodes, creating a star-like structure

The degree assortativity plot shows the relationship between a node's degree and the average degree of its neighbors. A positive trend line indicates assortative mixing.

### Modularity and Community Detection

**Modularity** measures how well a network can be divided into communities:
- **Formula:** M = Σ[Lc/L - (kc/2L)²]
  - Lc = edges within community c
  - L = total edges
  - kc = sum of degrees in community c

**Modularity values:**
- **Positive (> 0):** Communities are more connected internally than expected by chance
- **Higher values:** Stronger community structure
- **Typical range:** 0.3-0.7 for good community detection

**Results from Assignment 2:**
- Genre-based modularity: 0.0878 (weak community structure)
- Louvain algorithm modularity: 0.3187 (much stronger)
- Improvement: +0.2309

**Key Insight:** Structural communities (detected by algorithms) often capture relationships beyond simple categorical labels (like genres). Network structure reveals hidden patterns that categories alone cannot explain.

**Application:** In the Warhammer 40k network, faction affiliations might show similar patterns - characters may form communities based on narrative relationships rather than just faction membership.

### Network Analysis Best Practices

**1. Data Preprocessing:**
- Remove isolated nodes (degree = 0) for most analyses
- Consider keeping largest connected component only
- Filter low-degree nodes if needed (e.g., degree >= 2)

**2. Centrality Measures:**
- **Degree centrality:** Simple, measures direct connections
- **Betweenness centrality:** Identifies bridges/brokers in network
- **Eigenvector centrality:** Measures influence through connections to influential nodes
- **Closeness centrality:** Measures average distance to all other nodes

**3. Network Visualization:**
- Use appropriate layouts (ForceAtlas2, spring layout, etc.)
- Size nodes by degree or centrality
- Color by attributes (faction, community, etc.)
- Extract backbone for cleaner visualization

**4. Community Detection:**
- Compare detected communities to known categories (factions, genres)
- Use modularity to evaluate community quality
- Consider multiple algorithms (Louvain, Leiden, Infomap)

**5. Interpretation:**
- Consider network context (fictional universe, real-world relationships)
- Look for both expected and unexpected patterns
- Combine quantitative metrics with domain knowledge
